# 01 — Data Exploration

Explore the raw corpus collected by `scripts/make_dataset.py`.

**Paper Trail** | *The Documents Don't Lie*

```python
import json
import pandas as pd
from pathlib import Path

CORPUS_PATH = Path('../data/raw/raw_corpus.jsonl')

records = []
if CORPUS_PATH.exists():
    with CORPUS_PATH.open() as f:
        for line in f:
            if line.strip():
                records.append(json.loads(line))

df = pd.DataFrame(records)
print(f'Total records: {len(df)}')
df.head()
```

```python
# Source distribution
if not df.empty:
    df['source'].value_counts().plot(kind='bar', color=['#e63946','#f4a261','#e9c46a','#6c757d'],
                                      title='Documents by Source')
```

```python
# Text length distribution
if not df.empty:
    df['text_len'] = df['text'].str.len()
    df['text_len'].describe()
```
