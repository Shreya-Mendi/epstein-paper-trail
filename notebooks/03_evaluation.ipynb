# 03 — Evaluation

Review evaluation metrics: F1 scores, RAG Precision@5, hallucination rate, and robustness experiment.

Run `scripts/evaluate.py` first.

```python
import json
from pathlib import Path
from IPython.display import Image

REPORT_PATH = Path('../data/outputs/evaluation_report.json')
PLOT_PATH = Path('../data/outputs/robustness_experiment.png')

if REPORT_PATH.exists():
    with REPORT_PATH.open() as f:
        report = json.load(f)
    
    print('=== F1 Scores ===')
    for model, f1 in report.get('f1_scores', {}).items():
        print(f'  {model:30s}: {f1:.4f}')
    
    print(f"\nRAG Precision@5:    {report.get('rag_precision_at_5', 'N/A'):.4f}")
    print(f"Hallucination Rate: {report.get('hallucination_rate', 'N/A'):.4f}")
    
    rob = report.get('robustness', {})
    if rob:
        print(f"\n=== Robustness ===")
        print(f"  Clean F1:  {rob.get('clean_f1', 0):.4f}")
        print(f"  Noisy F1:  {rob.get('noisy_f1', 0):.4f}")
        print(f"  F1 Drop:   {rob.get('f1_drop', 0):.4f}")
else:
    print('evaluation_report.json not found. Run scripts/evaluate.py first.')
```

```python
if PLOT_PATH.exists():
    Image(str(PLOT_PATH))
else:
    print('Robustness plot not found.')
```
