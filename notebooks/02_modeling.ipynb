```python
import json
from pathlib import Path

SUMMARY_PATH = Path('../data/outputs/model_summary.json')

if SUMMARY_PATH.exists():
    with SUMMARY_PATH.open() as f:
        summary = json.load(f)
    for model, report in summary.items():
        if report and 'weighted avg' in report:
            f1 = report['weighted avg']['f1-score']
            print(f'{model:30s}  weighted F1: {f1:.4f}')
else:
    print('model_summary.json not found. Run scripts/model.py first.')
```
